```python
#extract file from zip file
import zipfile
zip_ref = zipfile.ZipFile("retail-orders.zip") 
zip_ref.extractall() # extract file to dir
zip_ref.close() # close file
```

```python
# read data from the file and  handle null value
import pandas as pd
df = pd.read_csv('orders.csv',na_values=['Not Available','unknown'])
df
```

```python
# check distinct values
df['Ship Mode'].unique()
```

```python
# rename columns names ..  make them lower case and replace sapece with underscore
df.rename(columns={'Order Id':'order_id','City':'city'})
# or
df.columns = df.columns.str.lower() # with this method we can convert all columns names into lowercase
```

```python
df.columns= df.columns.str.replace(' ','_') # with this method we can convert blank space with underscore
df.columns
```

```python
# dervive new columns discount , sale price and profit
# 1. discount column >> list_price * discount_percent/100
df['discount'] = df['list_price']*df['discount_percent']/100 
df['discount']
```

```python
# 2. sale column >> list_price - discount
df['sale_price'] = df['list_price']-df['discount']
df
```

```python
# 3. profit >> sale_price - cost_price
df['profit'] = df['sale_price'] - df['cost_price']
df['profit']
```

```python
# convert order date from object data type to datetime
df['order_date']= pd.to_datetime(df['order_date'],format='%Y-%m-%d')# this is the format of ur table date colum
df.dtypes
```

```python
# dropping unnessary  columns like discount_percent,cost_price,list_price
df.drop(columns=['list_price','discount_percent','cost_price']) # after it will still show in df 
# to avoid this we have to use inplace = true
df.drop(columns=['list_price','discount_percent','cost_price'],inplace=True)
```

```python
# note
'''
In Pandas, inplace=True means that changes will be applied directly to the DataFrame without creating a new copy. Let’s break it down!

📝 What Happens Without inplace=True?
By default, Pandas functions like drop(), fillna(), or sort_values() return a new DataFrame with changes applied, leaving the original DataFrame unchanged.
'''
```
